- Series: kiểu dữ liệu một chiều
- Index: dùng để biểu thị cho chỉ mục từ những hàng hay cột trong DataFrame.

```python
s = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
print(s)
```

DataFrame: kiểu dữ liệu chính trên Pandas và dùng để tượng trưng một bảng dữ liệu

```python
ten = pd.Series(["An", "Binh", "Cuong"])
tuoi = pd.Series([20, 21, 19])

df = pd.DataFrame({
    "Ten": ten,
    "Tuoi": tuoi,
})
print(df)
```

```python
s1 = pd.Series({"Ten" : "Duy ", "Tuoi" : 23})
s2 = pd.Series({"Ten" : "Chinh ", "Tuoi" : 24})
df = pd.DataFrame([s1, s2])
print(df)
```

```python
data = {
    "Ten" : ["Duy ","Binh", "Cuong"],
    "Tuoi" : [23, 25, 26],
    "Diem" : [10,9,8]
}
df = pd.DataFrame(data)
print(df)
```

```python
# 1. Tạo cột is_child: True nếu Age < 12, ngược lại False
t = pd.read_csv("Titanic.csv")
t["is_child"] = t["age"] < 12
```

```python
# 2. Lấy hành khách là nam (Sex == 'male') và là trẻ con (is_child == True)
t[(t["sex"] == "male") & (t["is_child"] == True)]
```

```python
# 3. Lấy tất cả hành khách > 60 tuổi
t[t["age"] > 60 ]
```

```python
# 4. Tính trung bình độ tuổi của hành khách (không tính giá trị thiếu)
# Thay thế NaN bằng giá trị trung bình
# Tính tuổi trung bình (bỏ qua NaN)
# t.fillna(0)
t = pd.read_csv("Titanic.csv")
mean = t['age'].mean()
t['age'] =t["age"].fillna(mean)
t
```

```python
# 6. Tìm hành khách ở tầng vé hạng nhất (Pclass == 1) và tuổi < 18
t[(t["pclass"] == 1) & (t["age"] < 18)]
```

```python
# 7. Tính số lượng hành khách ở mỗi hạng vé (Pclass)
t['pclass'].value_counts()
```

```python
# 8. Tính tỷ lệ sống sót (survived) theo từng hạng vé (pclass)
# a = t[(t["pclass"] == 1) & (t["survived"] == 1)]
# a['pclass'].value_counts()
#
# mean = a["count"] / t["count"]
t.groupby("pclass")["survived"].mean()
```

```python
# 1. Thêm cột FamilySize = SibSp + Parch
#  2. Thêm cột IsAlone:
# Nếu FamilySize == 0 thì là 1 (đi một mình),
#
# Ngược lại là 0.
# 3. Tách họ (LastName) từ cột Name và lưu vào cột LastName

# Cột Name có dạng: "LastName, Title Firstname"
# Dùng split để lấy phần trước dấu ","
```

```python
import pandas as pd

file = pd.read_csv("Titanic.csv")
# 1. Thêm cột FamilySize = SibSp + Parch
file["FamilySize"] = file["sibsp"] + file["parch"]
#  2. Thêm cột IsAlone:
# Nếu FamilySize == 0 thì là 1 (đi một mình),
# Ngược lại là 0.
file["IsAlone"] =(file["FamilySize"] == 0).astype(int)

# 3. Tách họ (LastName) từ cột Name và lưu vào cột LastName
# Cột Name có dạng: "LastName, Title Firstname"
# Dùng split để lấy phần trước dấu ","
file["LastName"] = file["name"].str.split(",", n=1, expand= True)[0]
file

```
