```python
import matplotlib.pyplot as plt
import numpy as np

xpoint = np.array([3,4,4,3,2,2,3,3])

ypoint = np.array([1,2,9,7,9,2,1,7])

plt.plot(xpoint,ypoint,marker='o',linestyle='dotted',color='violet',markersize=12,linewidth=5)
font1 = {"color" : "red"}
font2 = {"color" : "blue"}
plt.xlabel('x',font1,loc="left")
plt.ylabel('y',font2,loc="bottom")
plt.title('Data',fontsize=20)
plt.grid(color='r',linestyle='-',linewidth=0.2)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.subplot(1,2,1)
x1 = np.array([0,1,3,4])
y1 = np.array([4,4,10,2])
plt.plot(x1,y1,"o")

plt.subplot(1,2,2)
x2 = np.array([1,1,3,5])
y2 = np.array([3,1,1,9])
plt.plot(x2,y2,"o")


plt.suptitle('Data',fontsize=20)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.subplot(1,2,1)
x1 = np.array([0,1,3,4])
y1 = np.array([4,4,10,2])
plt.scatter(x1,y1,"o")

plt.subplot(1,2,2)
x2 = np.array([1,1,3,5])
y2 = np.array([3,1,1,9])
plt.scatter(x2,y2,"o")


plt.suptitle('Data',fontsize=20)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np



x2 = np.random.randint(100,size=90)
y2 = np.random.randint(100,size=90)

colors = np.random.randint(100,size=90)

size_point = 10*np.random.randint(100,size=90)
plt.scatter(x2,y2,c=colors,s=size_point,alpha=0.6)

plt.colorbar() #dai mau
plt.title('Data',fontsize=20)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x1 = np.array(["A","B","C","D"])
y1 = np.array([5,4,10,6])
plt.bar(x1,y1,color ="red")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

y=np.array([35,25,25,15])
mylabels = ["A","B","C","D"]
myexplode = [0,0.1,0.2,0.3]
mycolor = ["red","blue","green","yellow"]

plt.pie(y,labels=mylabels,colors=mycolor,explode=myexplode)
plt.legend(title = "Data",loc="best",bbox_to_anchor=(1,1))
plt.show()
```
