```python
# f(x) = 2*x + 5cos(x)
import numpy as np
import matplotlib.pyplot as plt
x = 5
epochs = 0
lr = 0.001
threshold = 0.00001 # ngưỡng / tìm giá trị tuyệt đối f(x) nhỏ hơn threshold

def f(x:float):
    return 2*x + 5*np.cos(x)

def f_(x:float):
    return 2 - 5*np.sin(x)

list_x = []

while abs(f(x)) > threshold:
    x = x - lr * f(x)
    epochs += 1
    list_x.append(x)
list_epochs = list(range(1,epochs+1))


plt.plot(list_epochs, list_x, label="f(x)")
plt.axis([-1,3500,-2,6])
plt.xlabel("epochs")
plt.ylabel("x(loss)")

plt.show()
```
