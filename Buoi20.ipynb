```python
from tarfile import version

import numpy as np
import matplotlib.pyplot as plt
x = np.array([[147,150,153,158,163,165,168,170,175,178,180,183]]).T
y = np.array([[49,50,51,54,58,58,60,62,63,66,67,68]]).T
# y = np.array(["Duy", "Chinh","Dung","Van","Que"])
```

```python
plt.plot(x,y,"ro")
plt.xlabel("Chieu Cao(cm)")
plt.ylabel("Can Nang(kg)")

plt.show()
```

```python
one = np.ones((x.shape[0],1))
xbar = np.concatenate((one,x),axis=1)
xbar
```

```python
# Nomarl Equation
a = np.dot(xbar.T,xbar)
b = np.dot(xbar.T,y)
w = np.dot(np.linalg.inv(a),b)
```

```python
#y = w1x + w0
print(f"y={x[1][0]}*x + {w[0][0]}")
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression(fit_intercept=False)
model.fit(xbar,y)

print(model.coef_)
```

```python
predict = model.coef_[0][0] + model.coef_[0][1]*183
print(predict)
```
