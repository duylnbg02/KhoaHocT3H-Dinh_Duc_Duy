```python
import numpy as np

x = np.array([[147,150,153,158,163,165,168,170,175,178,180,183]]).T
y = np.array([[49,50,51,54,58,58,60,62,63,66,67,68]]).T
one = np.ones((x.shape[0],1))
xbar = np.concatenate((one,x),axis=1)

class LinearRegressionCustom:
    def __init__(self, lr = 0.00001, epoch: int=100):
        self.lr = lr
        self.epochs = epoch
        self.weights = None
        self.bias = 0
        self.losses = []

    def fit(self, xbar, y, bias=0):
        n_sample, n_features = xbar.shape
        self.weights = np.zeros((n_features,1))
        self.bias = bias

        for epoch in range(self.epochs):
            y_pred = np.dot(xbar,self.weights) + self.bias
            # print("Predict:\n",y_pred)  # y_pred la ma tran output sau moi vong lap dung de so sanh voi y
            # print("Label:\n",y) # y la ma tran ban dau
            # print("Res:\n",y_pred - y) # y - y_pred
            # break
            dw = (1/n_sample)*np.dot(xbar.T,(y_pred - y))
            db = (1/n_sample)*np.sum(y_pred - y)

            self.weights = self.weights - self.lr * dw
            self.bias = self.bias - self.lr * db

            loss = self.loss_function_mse(y_pred, y)
            self.losses.append(loss)
            print(f"Epoch: {epoch}    Loss: {loss}  Bias: {self.bias} acc: {self.accuracy(xbar,y)}")


    @staticmethod
    def loss_function_mse(y_pred,y_):
        return np.mean((y_pred-y_)**2)

    def predict(self, x):
        return np.dot(x,self.weights)+self.bias

    def accuracy(self, x, y, threshold=3): # Tinh do chinh xac cua mo hinh
        y_pred = self.predict(x)
        y_pred_binary = (abs(y_pred-y) <= threshold).astype(int)
        correct_pred = np.sum(y_pred_binary)
        accuracy = correct_pred / len(y)
        return accuracy

model = LinearRegressionCustom(epoch=500)
model.fit(xbar,y)
test = np.array([1,150])
print(model.predict(xbar))
```
